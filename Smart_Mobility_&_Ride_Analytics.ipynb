### 1. Does trip distance significantly affect fare amount

### Linear Algebra Interpretation:

*   **Pearson Correlation**: In linear algebra, the Pearson correlation coefficient can be understood as the cosine of the angle between two centered vectors (i.e., vectors where their respective means have been subtracted). If we consider the `Trip_Distance` and `Fare_Amount` as vectors in a high-dimensional space, after centering them, a correlation close to 1 (or -1) implies that these vectors are nearly collinear (pointing in the same or opposite direction), indicating a strong linear relationship. A correlation close to 0 means the vectors are nearly orthogonal, suggesting little to no linear relationship.

*   **Covariance**: Covariance, in a linear algebra context, indicates how much two variables change together. If we think of `Trip_Distance` and `Fare_Amount` as random variables, their covariance measures the extent to which they deviate from their means in a similar fashion. A positive covariance indicates that when one variable increases, the other tends to increase as well (and vice-versa for negative covariance). In terms of vectors, it can be seen as a scaled dot product of the centered vectors, indicating the extent of their joint variability.

```python
# 1. Pearson Correlation
pearson_corr = df['Trip_Distance'].corr(df['Fare_Amount'])
print(f"Pearson Correlation between Trip_Distance and Fare_Amount: {pearson_corr:.2f}")

# 2. Covariance
covariance = df['Trip_Distance'].cov(df['Fare_Amount'])
print(f"Covariance between Trip_Distance and Fare_Amount: {covariance:.2f}")
```

#2. Are premium rides statistically more expensive than economy rides?

*  **Independent t-test**

An Independent t-test is used to compare the means of two independent groups.

*  **ANOVA**

ANOVA (Analysis of Variance) is used when we want to compare the means of three or more groups.


```python
from scipy import stats

# Filter data for Premium and Economy rides
premium_fares = df[df['Ride_Category'] == 'Premium']['Fare_Amount']
economy_fares = df[df['Ride_Category'] == 'Economy']['Fare_Amount']

# 1. Summarize with descriptive statistics
print("Descriptive Statistics for Premium Rides:")
print(premium_fares.describe())
print("\nDescriptive Statistics for Economy Rides:")
print(economy_fares.describe())

# 2. Compare groups using t-test
# Perform independent samples t-test
t_stat, p_value = stats.ttest_ind(premium_fares, economy_fares)

print(f"\nIndependent Samples t-test results:")
print(f"  t-statistic: {t_stat:.2f}")
print(f"  p-value: {p_value:.3f}")

# Interpret the p-value
alpha = 0.05
if p_value < alpha:
    print("  Conclusion: Reject the null hypothesis. There is a statistically significant difference in fare amounts between Premium and Economy rides.")
else:
    print("  Conclusion: Fail to reject the null hypothesis. There is no statistically significant difference in fare amounts between Premium and Economy rides.")
```

### 3. Is there a relationship between surge pricing and customer ratings?

```python
from scipy.stats import spearmanr

# Calculate Spearman correlation
spearman_corr, p_value_spearman = spearmanr(df['Surge_Multiplier'], df['Customer_Rating'])

print(f"Spearman Correlation between Surge_Multiplier and Customer_Rating: {spearman_corr:.2f}")
print(f"P-value: {p_value_spearman:.3f}")

# Interpret the p-value
alpha = 0.05
if p_value_spearman < alpha:
    print("\nConclusion: Reject the null hypothesis. There is a statistically significant monotonic relationship between surge pricing and customer ratings.")
else:
    print("\nConclusion: Fail to reject the null hypothesis. There is no statistically significant monotonic relationship between surge pricing and customer ratings.")
```

### 4. Do peak-hour rides differ in revenue compared to non-peak rides?

```python
# Filter data for Peak and Non-Peak rides
peak_fares = df[df['Ride_Time'] == 'Peak']['Fare_Amount']
non_peak_fares = df[df['Ride_Time'] == 'Non-Peak']['Fare_Amount']

# 1. Summarize with descriptive statistics
print("Descriptive Statistics for Peak Rides:")
print(peak_fares.describe())
print("\nDescriptive Statistics for Non-Peak Rides:")
print(non_peak_fares.describe())

# 2. Compare groups using t-test
# Perform independent samples t-test
t_stat_time, p_value_time = stats.ttest_ind(peak_fares, non_peak_fares)

print(f"\nIndependent Samples t-test results (Peak vs. Non-Peak):")
print(f"  t-statistic: {t_stat_time:.2f}")
print(f"  p-value: {p_value_time:.3f}")

# Interpret the p-value
alpha = 0.05
if p_value_time < alpha:
    print("  Conclusion: Reject the null hypothesis. There is a statistically significant difference in fare amounts between Peak and Non-Peak rides.")
else:
    print("  Conclusion: Fail to reject the null hypothesis. There is no statistically significant difference in fare amounts between Peak and Non-Peak rides.")
```

### 5. Are there unusual fare values that indicate anomalies or pricing errors?

```python
from scipy.stats import zscore

# 1. Detect outliers using IQR (Interquartile Range)
Q1 = df['Fare_Amount'].quantile(0.25)
Q3 = df['Fare_Amount'].quantile(0.75)
IQR = Q3 - Q1

lower_bound_iqr = Q1 - 1.5 * IQR
upper_bound_iqr = Q3 + 1.5 * IQR

outliers_iqr = df[(df['Fare_Amount'] < lower_bound_iqr) | (df['Fare_Amount'] > upper_bound_iqr)]

print(f"--- Outliers detected using IQR (1.5 * IQR criterion) ---")
print(f"Q1: {Q1:.2f}, Q3: {Q3:.2f}, IQR: {IQR:.2f}")
print(f"Lower Bound: {lower_bound_iqr:.2f}, Upper Bound: {upper_bound_iqr:.2f}")
print(f"Number of outliers: {len(outliers_iqr)}")
if not outliers_iqr.empty:
    print("Outlier Fare Amounts:")
    display(outliers_iqr[['Fare_Amount', 'Ride_Category', 'Trip_Distance']])
else:
    print("No outliers detected using IQR.")

# 2. Detect outliers using Z-score
df['Fare_Amount_Zscore'] = zscore(df['Fare_Amount'])

# A common threshold for outliers is a Z-score absolute value greater than 3
outliers_zscore = df[abs(df['Fare_Amount_Zscore']) > 3]

print(f"\n--- Outliers detected using Z-score (Absolute Z-score > 3) ---")
print(f"Number of outliers: {len(outliers_zscore)}")
if not outliers_zscore.empty:
    print("Outlier Fare Amounts (with Z-score):")
    display(outliers_zscore[['Fare_Amount', 'Fare_Amount_Zscore', 'Ride_Category', 'Trip_Distance']])
else:
    print("No outliers detected using Z-score.")

# Drop the temporary Z-score column if not needed further
df = df.drop(columns=['Fare_Amount_Zscore'])
```

### 8. Can statistical evidence support pricing or service improvements?

Statistical evidence can indeed support decisions regarding pricing and service improvements by quantifying uncertainty and validating assumptions. We will use confidence intervals and summarize hypothesis testing conclusions to provide actionable insights.

```python
import scipy.stats as st

# Calculate Confidence Interval for overall Fare_Amount
confidence_level = 0.95
mean_fare = df['Fare_Amount'].mean()
std_err_fare = st.sem(df['Fare_Amount'])

# Using t-distribution for small sample sizes or when population std dev is unknown
# (though N=200 is fairly large, t-dist is safer)
conf_int_fare = st.t.interval(confidence_level, len(df['Fare_Amount'])-1,
                               loc=mean_fare, scale=std_err_fare)

print(f"--- Confidence Interval for Overall Mean Fare Amount ({confidence_level*100:.0f}%) ---")
print(f"Mean Fare Amount: {mean_fare:.2f}")
print(f"Confidence Interval: ({conf_int_fare[0]:.2f}, {conf_int_fare[1]:.2f})")

# Calculate Confidence Intervals for mean Fare_Amount by Ride_Category
print("\n--- Confidence Intervals for Mean Fare Amount by Ride Category ---")
for category in df['Ride_Category'].unique():
    category_fares = df[df['Ride_Category'] == category]['Fare_Amount']
    if len(category_fares) > 1:
        mean_cat_fare = category_fares.mean()
        std_err_cat_fare = st.sem(category_fares)
        conf_int_cat_fare = st.t.interval(confidence_level, len(category_fares)-1,
                                          loc=mean_cat_fare, scale=std_err_cat_fare)
        print(f"{category:<10} Mean Fare: {mean_cat_fare:.2f}, CI: ({conf_int_cat_fare[0]:.2f}, {conf_int_cat_fare[1]:.2f})")
    else:
        print(f"{category:<10} Only one data point, cannot calculate CI.")

# Calculate Confidence Interval for overall Customer_Rating
print("\n--- Confidence Interval for Overall Mean Customer Rating ---")
mean_rating = df['Customer_Rating'].mean()
std_err_rating = st.sem(df['Customer_Rating'])
conf_int_rating = st.t.interval(confidence_level, len(df['Customer_Rating'])-1,
                               loc=mean_rating, scale=std_err_rating)
print(f"Mean Customer Rating: {mean_rating:.2f}")
print(f"Confidence Interval: ({conf_int_rating[0]:.2f}, {conf_int_rating[1]:.2f})")
```

### Bootstrapping for Mean Fare Amount

Bootstrapping is a resampling technique used to estimate the sampling distribution of a statistic by repeatedly drawing samples with replacement from the observed data. This allows us to construct robust confidence intervals without strong assumptions about the underlying data distribution.

```python
np.random.seed(42) # for reproducibility

n_iterations = 1000 # Number of bootstrap samples
bootstrap_means = []

for _ in range(n_iterations):
    # Resample with replacement
    bootstrap_sample = np.random.choice(df['Fare_Amount'], size=len(df['Fare_Amount']), replace=True)
    bootstrap_means.append(np.mean(bootstrap_sample))

# Calculate the 95% bootstrapped confidence interval
lower_bound = np.percentile(bootstrap_means, 2.5)
upper_bound = np.percentile(bootstrap_means, 97.5)

print(f"--- Bootstrapped Confidence Interval for Overall Mean Fare Amount ({confidence_level*100:.0f}%) ---")
print(f"Bootstrapped Mean: {np.mean(bootstrap_means):.2f}")
print(f"Bootstrapped Confidence Interval: ({lower_bound:.2f}, {upper_bound:.2f})")
```
